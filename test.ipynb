```python
def cal_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['price']*item['quantity']
    return total_cost

cart=[
    {'name':'apple','price':3,'quantity':7},
    {'name':'banana','price':4,'quantity':9},
    {'name':'orange','price':6,'quantity':8}
]

print(cal_total_cost(cart))
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficent balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")
        
    def get_balance(self):
        return self.balance

account=BankAccount("Adithya",20000)
print(account.balance)

account.deposit(2500)
account.withdraw(1500)
account.get_balance()
```

```python
class shape:
    def area(self):
        return "The area of figure"

class rectangle(shape):
    def __init__(self,width,length):
        self.width=width
        self.length=length

    def area(self):
        return self.width*self.length

class circle(shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius

def shape_area(shape):
    print(f"Area is {shape.area()}")

rectangle=rectangle(1,2)
circle=circle(2)

shape_area(rectangle)
shape_area(circle) 
```

```python
class Error(Exception):
    pass
class dobException(Error):
    pass

year = int(input("Enter the dob"))
age=2026-year
print(age)

try:
    if age<=30 and age>=20:
        print("The age is valid so you can apply for the exam")
    else:
        raise dobException
except dobException:
    print("Sorry your age is not within the limit")
```
